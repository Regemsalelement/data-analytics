## Data cleaning

### Subtask:
Clean the soil measurement data by handling potential outliers and ensuring data consistency.


**Reasoning**:
Identify potential outliers in numerical features ('N', 'P', 'K', 'ph') using boxplots and IQR, and handle them by capping at the 95th percentile.  Also, check for and correct inconsistencies in the 'crop' column.



```python
import matplotlib.pyplot as plt

# Identify outliers using boxplots and IQR for numerical features
numerical_cols = ['N', 'P', 'K', 'ph']
for col in numerical_cols:
    plt.figure(figsize=(6, 4))
    df.boxplot(column=col)
    plt.title(f'Boxplot of {col}')
    plt.show()

    # Cap outliers at the 95th percentile
    q1 = df[col].quantile(0.05)
    q3 = df[col].quantile(0.95)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

# Check for and correct inconsistencies in the 'crop' column
print("Unique crop values before cleaning:", df['crop'].unique())
# Assuming valid crop names are already known or can be determined from domain knowledge
# Replace any typos or inconsistent values with the correct names
# Example:
# df['crop'] = df['crop'].replace({'ricee': 'rice', 'maizee': 'maize'})

# Convert 'crop' column to categorical data type
df['crop'] = pd.Categorical(df['crop'])

# Re-examine data distributions after cleaning
for col in numerical_cols:
    plt.figure(figsize=(6, 4))
    df.boxplot(column=col)
    plt.title(f'Boxplot of {col} after cleaning')
    plt.show()
```

## Data preparation

### Subtask:
Prepare the data for clustering by standardizing numerical features and one-hot encoding the categorical feature 'crop'.


**Reasoning**:
Prepare the data for clustering by standardizing numerical features and one-hot encoding the categorical feature 'crop'.



```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import pandas as pd
import numpy as np

# Separate numerical and categorical features
numerical_features = df[['N', 'P', 'K', 'ph']]
categorical_feature = df[['crop']]

# Standardize numerical features
scaler = StandardScaler()
scaled_numerical = scaler.fit_transform(numerical_features)
scaled_numerical = pd.DataFrame(scaled_numerical, columns=['N', 'P', 'K', 'ph'])

# One-hot encode categorical feature
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded_crop = encoder.fit_transform(categorical_feature)
encoded_crop_df = pd.DataFrame(encoded_crop, columns=encoder.get_feature_names_out(['crop']))

# Concatenate the standardized numerical features and the one-hot encoded 'crop' data
df_prepared = pd.concat([scaled_numerical, encoded_crop_df], axis=1)
display(df_prepared.head())
```

## Data clustering

### Subtask:
Perform K-means clustering on the prepared data.


**Reasoning**:
Perform K-means clustering on the prepared data, add cluster assignments as a new column, and visualize the clusters using PCA for dimensionality reduction.



```python
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Initialize KMeans model
kmeans = KMeans(n_clusters=5, n_init=10, random_state=0)

# Fit the model to the prepared data
kmeans.fit(df_prepared)

# Add cluster assignments to the dataframe
df_prepared['cluster'] = kmeans.labels_

# Reduce dimensionality using PCA for visualization
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_prepared.drop('cluster', axis=1))
df_pca = pd.DataFrame(df_pca, columns=['PC1', 'PC2'])
df_pca['cluster'] = df_prepared['cluster']

# Visualize the clusters
plt.figure(figsize=(8, 6))
for cluster in df_pca['cluster'].unique():
    plt.scatter(df_pca[df_pca['cluster'] == cluster]['PC1'],
                df_pca[df_pca['cluster'] == cluster]['PC2'],
                label=f'Cluster {cluster}')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-means Clustering Visualization (PCA)')
plt.legend()
plt.show()
```

## Model evaluation

### Subtask:
Evaluate the K-means clustering performance using the Silhouette score and Davies-Bouldin Index.


**Reasoning**:
Calculate the Silhouette score and Davies-Bouldin Index to evaluate the KMeans clustering performance.



```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Calculate Silhouette score
silhouette_avg = silhouette_score(df_prepared.drop('cluster', axis=1), df_prepared['cluster'])
print(f"Silhouette Score: {silhouette_avg}")

# Calculate Davies-Bouldin Index
davies_bouldin_index = davies_bouldin_score(df_prepared.drop('cluster', axis=1), df_prepared['cluster'])
print(f"Davies-Bouldin Index: {davies_bouldin_index}")
```
