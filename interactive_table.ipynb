```python
import pandas as pd
import numpy as np
import ipywidgets as widgets
from IPython.display import display

# Assuming 'data' is your DataFrame containing the soil measures

def filter_data(N, P, K, ph):
    filtered_data = data[
        (data['N'] >= N[0]) & (data['N'] <= N[1]) &
        (data['P'] >= P[0]) & (data['P'] <= P[1]) &
        (data['K'] >= K[0]) & (data['K'] <= K[1]) &
        (data['ph'] >= ph[0]) & (data['ph'] <= ph[1])
    ]
    display(filtered_data)

# Create sliders for each parameter
N_slider = widgets.IntRangeSlider(min=data['N'].min(), max=data['N'].max(), step=1, value=[data['N'].min(), data['N'].max()], description='N:')
P_slider = widgets.IntRangeSlider(min=data['P'].min(), max=data['P'].max(), step=1, value=[data['P'].min(), data['P'].max()], description='P:')
K_slider = widgets.IntRangeSlider(min=data['K'].min(), max=data['K'].max(), step=1, value=[data['K'].min(), data['K'].max()], description='K:')
ph_slider = widgets.FloatRangeSlider(min=data['ph'].min(), max=data['ph'].max(), step=0.1, value=[data['ph'].min(), data['ph'].max()], description='ph:')


# Interactive output
widgets.interactive(filter_data, N=N_slider, P=P_slider, K=K_slider, ph=ph_slider)
```

The data appears to contain soil measurements (N, P, K, and pH) alongside crop information. The columns 'crop' and 'урожай' appear to be duplicates, as "урожай" is the Russian word for "crop." The data is already in the correct format, and the columns N, P, K, and ph are already of type int64 and float64 as needed.
